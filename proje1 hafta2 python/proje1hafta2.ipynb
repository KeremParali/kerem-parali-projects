```python
"""
KEREM PARALI - 300125024

PROJE 1: HAFTA 2 - Veri Seti Seçimi ve Doğru Okuma
Spotify Tracks veri setinin okunması, kodlaması (utf-8),
boyutları ve sütun veri tiplerinin (dtypes) incelenmesi.
"""
import pandas as pd

# Dosyayı tanımlayıp yüklüyoruz
dosya = "spotify-tracks-dataset.csv"

# 1. VERİ SETİNİ OKUMA VE KODLAMA (ENCODING)
# Türkçe/özel karakterlerin bozulmaması için utf-8 ile açıyoruz
try:
    df = pd.read_csv(dosya, encoding="utf-8")
    print("-> Veri seti UTF-8 ile başarıyla yüklendi.\n")
except Exception as e:
    print("Okuma hatası oluştu:", e)

# 2. BOYUT İNCELEMESİ (SHAPE)
# Satır ve sütun sayılarına bakıyoruz
print("=" * 40)
print("1. VERİ SETİ BOYUT BİLGİLERİ")
print("=" * 40)
print(f"Satır Sayısı  : {df.shape[0]:,}")
print(f"Sütun Sayısı  : {df.shape[1]}")
print(f"Toplam Hücre  : {df.size:,}\n")

# 3. VERİ TİPLERİ İNCELEMESİ (DATA TYPES)
# Sütun tiplerini (int, float, object vb.) kontrol ediyoruz
print("=" * 40)
print("2. SÜTUN VERİ TİPLERİ")
print("=" * 40)
print(df.dtypes)
print("\n")

# 4. GENEL TABLO ÖZETİ (INFO)
# Boş değerler ve bellek kullanımı için genel özet
print("=" * 40)
print("3. GENEL VERİ ÖZETİ")
print("=" * 40)
df.info()
```
