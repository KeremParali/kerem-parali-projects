```python
import numpy as np 
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression

# Çalışma saatleri (X)
x = np.array([1, 2, 3, 4, 5]).reshape(-1, 1)

# Sınav notları (Y)
y = np.array([45, 50, 60, 65, 75])

# Regresyon modeli oluşturma
model = LinearRegression()

# Modeli eğitme
model.fit(x, y)

# Tahmin yapma
tahmin = model.predict([[6]])

print("6 saat çalışan öğrencinin tahmini notu:", tahmin[0])

# Grafik oluşturma
plt.scatter(x, y, color='blue', label='Gerçek Veriler')

# Regresyon doğrusunu çizme
plt.plot(x, model.predict(x), color='red', label='Regresyon Doğrusu')

# Grafik başlığı ve eksen isimleri
plt.title("Regresyon Analizi")
plt.xlabel("Çalışma Saati")
plt.ylabel("Sınav Notu")

# Açıklama kutusu
plt.legend()

# Göster
plt.show()
```
