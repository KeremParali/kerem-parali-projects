```python
"""
PYTHON DÖNEM PROJESİ
Nesneye Yönelik Veri Ön İşleme Aracı
KEREM PARALI
NO:300125024
Bu projede Titanic veri seti kullanılarak,
veri temizleme ve analiz işlemleri OOP mantığıyla yapılmıştır.
(analiz.ipynb)
"""
# Kütüphanedeki DataPreprocessor sınıfını içe aktar
from kutuphane import DataPreprocessor

# Titanic veri setini yükle
verisetim = DataPreprocessor("veri/titanic.csv")

# Veri setinin genel özetini göster
verisetim.veri_ozeti()

# Eksik verileri kontrol et
verisetim.eksik_veri_kontrolu()

# Eksik verileri ortalama ile doldur
verisetim.eksikleri_doldur(strateji='mean')

# Gereksiz sütunu sil
verisetim.sutun_sil("PassengerId")

# Age sütununun dağılım grafiğini çiz
verisetim.dagilim_ciz("Age")

# Temizlenmiş veriyi yeni CSV dosyası olarak kaydet
verisetim.veriyi_kaydet("temiz_veri.csv")
```
